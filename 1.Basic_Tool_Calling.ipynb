## OpenAI library to make local and gpt model have a conversation between each other

```python
import os
import requests
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
load_dotenv(override=True)
openai_api_key = os.getenv('OPENAI_API_KEY')
google_api_key = os.getenv('GOOGLE_API_KEY')

# Verify keys are loaded and not empty
if openai_api_key:
    print(f"OpenAI key loaded: {openai_api_key[:10]}...{openai_api_key[-4:]}")
else:
    print("OpenAI key not found")

if google_api_key:
    print(f"Google key loaded: {google_api_key[:10]}...{google_api_key[-4:]}")
else:
    print("Google key not found")
```

```python
ollama_url = "http://localhost:11434/v1"
gemini_url = "https://generativelanguage.googleapis.com/v1beta/openai/"


openai = OpenAI(api_key=openai_api_key)
gemini = OpenAI(api_key=google_api_key, base_url=gemini_url)
ollama = OpenAI(api_key="ollama", base_url=ollama_url)
```

```python
tell_a_joke = [
    {"role": "system", "content":"You are a snarky stand up comedian"},
    {"role": "user", "content": "Tell a joke for a student on the journey to becoming an expert in LLM Engineering"},
    {"role":"assistant", "content":"Why did the LLM break up with the data scientist? Because it needed more than just a good prompt to feel connected!"}
]
```

```python
response = ollama.chat.completions.create(model="phi3", messages=tell_a_joke)
display(Markdown(response.choices[0].message.content))
```

```python
response = openai.chat.completions.create(model="gpt-4o", messages=tell_a_joke)
display(Markdown(response.choices[0].message.content))
```

```python
# Define different system prompts for each LLM
system_prompts = {
    "ollama": "You are a snarky, sarcastic comedian who loves making puns and isn't afraid to be a bit edgy.",
    "gemini": "You are a wholesome, cheerful friend who always looks on the bright side and tries to keep things positive."
}

# Initialize the conversation
conversation = [
    {"role": "user", "content": "Hey! Long time no see. How have you been?"}
]

# Keep track of which model said what (optional)
conversation_metadata = []

# Number of conversation turns
num_turns = 6

for i in range(num_turns):
    # Determine which model to use
    model_name = "ollama" if i % 2 == 0 else "gemini"
    
    # Prepare messages with the specific system prompt for this model
    messages_with_system = [{"role": "system", "content": system_prompts[model_name]}] + conversation
    
    # Get response from the appropriate model
    if model_name == "ollama":
        response = ollama.chat.completions.create(model="phi3", messages=messages_with_system)
        display_name = "LLM 1 (Ollama - Snarky)"
    else:
        response = openai.chat.completions.create(model="gpt-5-mini", messages=messages_with_system)
        display_name = "LLM 2 (Gemini - Wholesome)"
    
    # Get the assistant's response
    assistant_message = response.choices[0].message.content
    
    # Append to conversation
    conversation.append({"role": "assistant", "content": assistant_message})
    
    # Track metadata (optional)
    conversation_metadata.append({"model": model_name, "content": assistant_message})
    
    # Display the response
    display(Markdown(f"**{display_name}:** {assistant_message}"))
    
    # Add a user message to continue
    conversation.append({"role": "user", "content": "Respond to what they said"})

# Display conversation with model labels
print("\n--- Conversation with Model Labels ---")
for i, meta in enumerate(conversation_metadata):
    print(f"{meta['model'].upper()}: {meta['content']}\n")
```
